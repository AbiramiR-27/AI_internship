## PROS AND CONS OF MODNET

Pros
* Lightweight and efficient architecture
* Supports real-time portrait matting
* Produces high-quality alpha mattes
* Preserves fine details such as hair and object boundaries
* Suitable for mobile and edge devices
* Requires fewer computational resources compared to heavier matting models

Cons
* Primarily designed for human portrait segmentation
* Less effective for general object segmentation tasks
* Performance depends on the quality and diversity of portrait datasets
* May struggle with complex backgrounds and unusual poses
* Not optimized for multi-object segmentation
